## Práctica
En equipos de 3 crear un programa que evalue un funcion de utilidad de chelería de tepito cuyo producto estrella vale 90 pesos.

Tiene que pagar mensualmente 45,000 de renta, 102,000 de nomina y 32,000 de servicios. Cada michelada requiere 1 caguama de 35 pesos, vaso 2 pesos, escarchado y gomitas 3.5 pesos, servilleta .5 pesos, limón y salsas 4.3 pesos. El programa debe tener la opción de:

* Calcular un valor de utilidad de acuerdo a ciertas piezas vendidas. advirtiendo si es ganancia o pérdida.

* Mostrar el valor de utilidad dentro de un intervalo de piezas vendidas indicado por el usuario.

* Calcular el punto de equilibrio.

* Calcular la cantidad de piezas necesarias para obtener un objetivo financiero


$$B(x) = I(x) - C(x)$$
* $B(x) := $ Función beneficio / utilidad
* $I(x) : = $ Función de ingresos totales
* $C(x) := $ Función de costos totales

```python
c1, c2, c3 = 45000, 102000, 32000 # establecemos costos de operación
p1, p2, pp3, p4, pp5 = 35, 2, 3.5, 1/2, 4.3 # establecemos costos unitario de producción
c = c1 + c2 +c3 # costos fijos al mes
P = p1 + p2 + pp3 + p4 + pp5 # costo por miche
V =  90 # precio de venta

print("\n" + "<"*57)
t = float(input('Ingrese la cantidad de meses del periodo: '))
print(">"*57)
c *= t
opc = 0
while opc != 5:
  print("\n" + "°"*57)
  print("      CHELERIA DE TEPITO - EVALUAR FUNCIÓN DE UTILIDAD")
  print("°"*57)
  print("1. Utilidad por piezas vendidas")
  print("2. Utilidad dentro de un intervalo de piezas vendidas")
  print("3. Calcular punto de equilibrio")
  print("4. Piezas necesarias para un objetivo financiero")
  print("5. Salir")

  opc = int(input("\nSelecciona una opción: "))

  # opcion 1
  if opc == 1:
     n = int(input("Ingresa el número de micheladas que vendiste al mes: "))
     I = V * n
     C = c + P * n
     B = I - C

     if B < 0:
      print(f"Con una venta de {n} micheladas, tuviste una PÉRDIDA de: ${-B:.2f}")
     elif B == 0:
      print(f"Punto de equilibrio alcanzado con {n} micheladas vendidas ($0.00)")
     else:
      print(f"Con una venta de {n} micheladas, tuviste una GANANCIA de: ${B:.2f}")

  # opcion 2
  elif opc == 2:
    inicio = int(input("Cantidad inicial del intervalo: "))
    fin = int(input("Cantidad final del intervalo: "))
    paso = int(input("Incremento/paso: "))
    print("\nPiezas | Utilidad / Pérdida")
    print("-" * 30)
    for n in range(inicio, fin + 1, paso):
        I = V * n
        C = c + P * n
        B = I - C
        print(f"{n:<7} | ${B:>13.2f}")

  # opcion 3
  elif opc == 3:
    n = 0
    I = V * n
    C = c + P * n
    while I < C:
      n += 1
      I = V * n
      C = c + P * n
    print(f"Debes vender mínimo {int(n)} micheladas al mes para no perder ni ganar.")
    print(f"Ingresos: ${I:.2f} | Costos: ${C:.2f}")

  # opcion 4
  elif opc == 4:
    meta = float(input("¿Cuál es tu objetivo de ganancia mensual? "))
    n = 0
    I = V * n
    C = c + P * n

    while (I-C) < meta :
      n += 1
      I = V * n
      C = c + P * n
      print(f"Para ganar ${meta:.2f}, necesitas vender {n} micheladas en el mes.")
      print(f"Ganancia exacta lograda: ${I - C:.2f}")

  # salir
  elif opc == 5:
        print("Programa finalizado")

  else:
        print("Opción inválida, intenta de nuevo.")
```
