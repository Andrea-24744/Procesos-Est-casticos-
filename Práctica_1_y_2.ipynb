```python
import random

calif = [random.randint(0, 100) for i in range(2000)]
claves = [chr(random.randint(65, 90)) + chr(random.randint(65, 90)) + chr(random.randint(65, 90)) + chr(random.randint(65, 90)) for i in range(2000)]

for i in range(2000):
  print(f"{i+1}. Clave: {claves[i]} - Calificación: {calif[i]}")

max_c = max(calif)
min_c = min(calif)
prom = sum(calif) / len(calif)
cien = calif.count(100)
mas_repetida = 0
max_frecuencia = 0

for c in set(calif):
  cant = calif.count(c)
  if cant > max_frecuencia:
    max_frecuencia = cant
    mas_repetida = c

print(f"Calificación más alta: {max_c}")
print(f"Calificación más baja: {min_c}")
print(f"Promedio: {prom:.2f}")
print(f"Veces que aparece el 100: {cien}")
print(f"Calificación que más se repite: {mas_repetida}")

# Aprobados
aprob_c = []
aprob_k = []
for i in range(len(calif)):
  if calif[i] >= 60:
    aprob_c.append(calif[i])
    aprob_k.append(claves[i])

prom_aprob = sum(aprob_c) / len(aprob_c)
print(f"Cantidad de aprobados: {len(aprob_c)}")
print(f"Promedio aprobados: {prom_aprob:.2f}")

# Reprobados
reprob_c = []
reprob_k = []

for i in range(len(calif)):
  if calif[i] < 60:
    reprob_c.append(calif[i])
    reprob_k.append(claves[i])

prom_reprob = sum(reprob_c) / len(reprob_c)
print(f"Cantidad de reprobados: {len(reprob_c)}")
print(f"Promedio reprobados: {prom_reprob:.2f}")

busqueda = input("Ingresa la clave a buscar: ")
if busqueda in claves:
  pos = claves.index(busqueda)
  print(f"La calificación de la clave {busqueda} es: {calif[pos]}")
else:
  print("La clave no existe")
```
